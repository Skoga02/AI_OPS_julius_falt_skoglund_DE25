# Chatting with pydantic ai

```python
from pydantic_ai import Agent
from model import nvidia_model

chat_agent= Agent(nvidia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
            instructions="Be a joking programing nerd, always answer with a programming joke. Also add in some emojis to make it fun.")

chat_agent
```

```python
result = await chat_agent.run("How is your day?")
result.output
```

```python
# NO MEMORY
result2 = await chat_agent.run("What did i just ask you?")
print(result2.output)
```

```python
result2.all_messages()
```

## Give a memory

```python
result = await chat_agent.run("Varför är tåget alltid sent i stockholm?")
print(result.output)
```

```python
result = await chat_agent.run("gissa på vad vädret är i här och säg vart här är", message_history=result.all_messages())
print(result.output)
```

```python
result.all_messages()
```

```python
# amount of messages
len(result.all_messages())
```

```python
# first message
result.all_messages()[0]
```

## Chat with our API endpoint

```python
import httpx

result = httpx.post(
    "http://127.0.0.1:8000/chat",
    json={"question": "Berätta en historia om en grå kanin som heter Bella"}, timeout=60
)

result
```

```python
result.json().keys()
```

```python
message_history = result.json().get("message_history")
message_history[0]
```

```python
print(result.json().get("message_history"))
```

```python
result = httpx.post(
    "http://127.0.0.1:8000/chat",
    json={"question": "I föregående historia, vad hette kaninen?", "message_history": message_history}, timeout=60
)

result
```

```python
print(result.json().get("response"))
```
