## Test OpenMeteo

```python
import httpx 

response = httpx.get(
    "https://api.open-meteo.com/v1/forecast",
    params={
        "latitude": 59.33,
        "longitude": 18.07,
        "current": "temperature_2m"
    },
)

response.json()
```

```python
response = httpx.get(
    "https://geocoding-api.open-meteo.com/v1/search",
    params={"name": "stockholm"}
)
results = response.json().get("results")
results
```

```python
response = httpx.get(
    "https://geocoding-api.open-meteo.com/v1/search",
    params={"name": "stockholm", "count":1}
)
results = response.json().get("results")
results
```
